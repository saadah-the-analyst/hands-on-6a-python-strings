# Hands-on Exercise 1

```python
# Method 1: String Concatenation (+)

# store patient registration details in variables
first_name = "Grace"
last_name = "Johnson"
age = 34
blood_group = "O+"
department = "Cardiology"

# build the patient card using string concatenation 
output = ("================= CAREPLUS HOSPITAL ==================\n"   
"Patient Name: " + first_name + " " + last_name + "\n"         
"Age: " + str(age) + " " + "Years\n"         
"Blood Group: " + blood_group + "\n"        
"Department: " + department + "\n"         
"======================================================")

# output the formatted registration card
print(output)
```

# Bonus Challenge

```python
# Method 2: .format() method

# store patient registration details in variables
first_name = "Grace"
last_name = "Johnson"
age = 34
blood_group = "O+"
department = "Cardiology"

# format the card using string placeholders {} and .format()
output = """================= CAREPLUS HOSPITAL ==================  
Patient Name: {} {}
Age: {} Years     
Blood Group: {}    
Department: {}      
"======================================================""".format(first_name, last_name, age, blood_group, department)# 

# output the formatted registration card
print(output)
```

```python
# Method 3: f-string Formatting

# store patient registration details in variables
first_name = "Grace"
last_name = "Johnson"
age = 34
blood_group = "O+"
department = "Cardiology"

# format the patient card using f-strings with variables inside {}
output = f"""================= CAREPLUS HOSPITAL ==================  
Patient Name: {first_name} {last_name}
Age: {age} Years     
Blood Group: {blood_group}    
Department: {department}      
"======================================================"""

# output the formatted registration card
print(output)
```

```python
# Method 4: f-string with Uppercase Name

# store patient registration details in variables
first_name = "Grace"
last_name = "Johnson"
age = 34
blood_group = "O+"
department = "Cardiology"

# use .upper() inside the f-string placeholders to display the full name in uppercase
output = f"""================= CAREPLUS HOSPITAL ==================  
Patient Name: {first_name.upper()} {last_name.upper()}
Age: {age} Years     
Blood Group: {blood_group}    
Department: {department}      
"======================================================"""

# output the formatted registration card
print(output)
```

# Hands-on Exercise 2

```python
log_message = "  LOGIN SUCCESSFUL FROM ATM-045 ON 2026-07-25  "

print("Original Log:",log_message)

cleaned_log = log_message.strip()
print("Clean Log:",cleaned_log)

lowercase_log = cleaned_log.lower()
print("Lowercase Log:",lowercase_log)

position = cleaned_log.find("SUCCESSFUL")
print("Position of SUCCESSFUL:",position)

updated_log = cleaned_log.replace("SUCCESSFUL", "VERIFIED")
print("Updated Log:",updated_log)
```

# Bonus Challenge

```python
# calculate the nunber of  characters before cleaning
lenght_before = len(log_message)
print("Characters before cleaning:",lenght_before)

# calculate the number of  characters after cleaniung 
lenght_after = len(cleaned_log)
print("Characters after cleaanung:",lenght_after)

# count the number of words in the cleaned log
word_count = len(cleaned_log.split())
print("Number of words in cleand log:",word_count)
```

# Hands-on Exercise 3

```python
# store the movie title in a variable and display it 
movie_title = "The Last Kingdom Returns"
print("Movie Title:",movie_title)

# print the first character
first_character = movie_title[0]
print("First Character:",first_character)

# print the last character
last_character = movie_title[-1]
print("Last Chsaracter:",last_character)

# print the first word only
first_word = movie_title[0:3]
print("First Word:",first_word)

# print the last word only
last_word = movie_title[-7:]
print("Last Word:",last_word)

# find the starting position of the word "Returns"
position = movie_title.find("Returns")
print("Position of 'Returns':",position)

# split the movie title into individual words
split_movie = movie_title.split()
print("Split Movie:",split_movie)
```

```python
# ask the user for a movie title
user_title = input("Enter any movie title:")

# display total number of words
word_count = len(user_title.split())
print("Total Words:",word_count)

# display the total number of characters 
character_count = len(user_title)
print("Total Characters:",character_count)

# convert title to title case 
title_case = user_title.title()
print("Title Case:",title_case)
```
